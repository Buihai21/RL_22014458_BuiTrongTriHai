## PHẦN B & C: THỐNG KÊ THỰC NGHIỆM TRÊN CARTPOLE-V1

```python
env = gym.make("CartPole-v1")
rewards = []
for ep in range(100):
    obs, info = env.reset(seed=ep)
    ep_reward = 0
    while True:
        action = env.action_space.sample()
        obs, reward, terminated, truncated, info = env.step(action)
        ep_reward += reward
        if terminated or truncated:
            break
    rewards.append(ep_reward)
env.close()

r_arr = np.array(rewards)
print(f"Mean reward : {np.mean(r_arr):.2f}")
print(f"Min reward  : {np.min(r_arr):.2f}")
print(f"Max reward  : {np.max(r_arr):.2f}")
print(f"Std reward  : {np.std(r_arr):.2f}")
```

## PHẦN D: SO SÁNH HIỆU NĂNG CÁC AGENT (BÀI 35)

```python
def random_policy(obs, env): return env.action_space.sample()
def angle_policy(obs, env): return 1 if obs[2] > 0 else 0
def improved_policy(obs, env): return 1 if (obs[2] + 0.15 * obs[3]) > 0 else 0

def evaluate(policy_func):
    env = gym.make("CartPole-v1")
    rewards = []
    for ep in range(500):
        obs, info = env.reset(seed=ep)
        r_sum = 0
        while True:
            action = policy_func(obs, env)
            obs, reward, terminated, truncated, info = env.step(action)
            r_sum += reward
            if terminated or truncated: break
        rewards.append(r_sum)
    env.close()
    return np.array(rewards)

r_rand = evaluate(random_policy)
r_angle = evaluate(angle_policy)
r_impr = evaluate(improved_policy)

print(f"{'Agent':<15} | {'Mean Reward':<12} | {'Std':<10} | {'Min':<8} | {'Max':<8} | {'Mean Length':<12}")
print("-" * 75)
for name, arr in [("Random", r_rand), ("Angle-based", r_angle), ("Improved", r_impr)]:
    print(f"{name:<15} | {np.mean(arr):>12.2f} | {np.std(arr):>10.2f} | {np.min(arr):>8.1f} | {np.max(r_arr):>8.1f} | {np.mean(arr):>12.2f}")
```

## PHẦN 10: CÂU HỎI VÀ CÂU TRẢ LỜI BÁO CÁO (THEO YÊU CẦU ĐỀ BÀI)

1. **Agent là gì?**
   - Agent (tác nhân) là thành phần học tập và đưa ra quyết định (hành động) dựa trên trạng thái quan sát được từ môi trường.

2. **Environment là gì?**
   - Environment (môi trường) là thế giới bên ngoài mà Agent tương tác. Môi trường nhận hành động từ Agent, cập nhật trạng thái và trả về quan sát mới cùng phần thưởng (reward).

3. **Observation khác action như thế nào?**
   - **Observation:** Thông tin/trạng thái của môi trường mà Agent nhận được tại thời điểm $t$.
   - **Action:** Hành động mà Agent chủ động lựa chọn để thực hiện và tác động ngược lại môi trường.

4. **Reward dùng để làm gì?**
   - Reward (phần thưởng) là tín hiệu phản hồi dạng số định lượng mức độ tốt/xấu của hành động mà Agent vừa thực hiện, giúp định hướng mục tiêu học tập của Agent.

5. **Episode là gì?**
   - Episode là một chuỗi tương tác hoàn chỉnh từ lúc khởi tạo môi trường (`reset`) cho tới khi gặp điều kiện kết thúc (`terminated` hoặc `truncated`).

6. **Policy là gì?**
   - Policy ($\pi$) là chiến lược chọn hành động của Agent, ánh xạ từ quan sát/trạng thái hiện tại $s$ sang hành động $a$.

7. **action_space thể hiện điều gì?**
   - `action_space` thể hiện tập hợp tất cả các hành động hợp lệ mà Agent có thể thực hiện trong môi trường.

8. **observation_space thể hiện điều gì?**
   - `observation_space` thể hiện cấu trúc, miền giá trị và kiểu dữ liệu của tất cả quan sát mà môi trường trả về cho Agent.

9. **reset() được gọi khi nào?**
   - `reset()` được gọi ở đầu mỗi episode mới để đưa môi trường về trạng thái khởi tạo ban đầu.

10. **step(action) thực hiện việc gì?**
    - `step(action)` áp dụng hành động `action` vào môi trường, biến đổi trạng thái sang thời điểm $t+1$ và trả về bộ 5 giá trị: `(observation, reward, terminated, truncated, info)`.

11. **terminated và truncated khác nhau thế nào?**
    - **Terminated:** Episode kết thúc do bản chất quy luật môi trường (ví dụ con lắc bị đổ trong CartPole, rớt xuống hố trong FrozenLake).
    - **Truncation:** Episode bị ngắt do giới hạn bên ngoài (ví dụ đạt giới hạn số bước tối đa 500 bước).

12. **Random policy có phải là một thuật toán học tăng cường không? Giải thích.**
    - Không. Random policy chỉ chọn ngẫu nhiên hành động mà không có quá trình cập nhật tri thức, không dựa trên reward thu được để tối ưu hóa quyết định trong tương lai.

13. **Vì sao phải chạy nhiều episode thay vì chỉ một episode?**
    - Vì môi trường RL và Policy có yếu tố ngẫu nhiên (stochastic). Chạy nhiều episode giúp tính các chỉ số thống kê (trung bình, độ lệch chuẩn) đánh giá độ ổn định của Agent một cách khách quan.

14. **Vì sao cần random seed?**
    - Random seed giúp cố định chuỗi sinh số ngẫu nhiên, đảm bảo tính tái lập (reproducibility) của các thí nghiệm khi chạy lại.

15. **Reward trung bình có ý nghĩa gì khi đánh giá agent?**
    - Reward trung bình phản ánh hiệu năng tổng thể và mức độ ổn định của Agent qua nhiều lần thử nghiệm. Chi phí reward trung bình càng cao chứng tỏ Agent học càng hiệu quả.
